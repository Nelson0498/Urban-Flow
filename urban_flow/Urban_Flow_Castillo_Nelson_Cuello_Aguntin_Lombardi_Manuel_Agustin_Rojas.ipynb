````python
# Título: Ejercicio 03 - Normalización de Fechas y Horas (Editado por Manu Lombardi)

def normalizar_datos_temporales(df: pd.DataFrame) -> pd.DataFrame:
  """
  Normaliza las columnas de fecha y hora del sistema heredado.
  Las fechas inválidas se establecen en 1932-01-01 y las horas en 00:00.

  Args:
    df (pd.DataFrame): DataFrame con columnas 'fecha' y 'hora'.

  Returns:
    pd.DataFrame: DataFrame con formatos temporales estandarizados.
  """

  fechas: pd.Series = pd.to_datetime(df['fecha'], errors='coerce', dayfirst=True)
  df['fecha'] = fechas.fillna(
    pd.Timestamp('1932-01-01')
  ).dt.strftime('%Y-%m-%d')

  print("Primeras 10 fechas normalizadas:")
  print(df['fecha'].head(10))
  print("-" * 40)
  print()


  horas: pd.Series = pd.to_datetime(
    df['hora'], format='mixed', errors='coerce'
  )
  df['hora'] = horas.fillna(
    pd.Timestamp('1900-01-01 00:00')
  ).dt.strftime('%H:%M')

  print("Primeras 10 horas normalizadas:")
  print(df['hora'].head(10))

  return df

df_fines_raw = normalizar_datos_temporales(df_fines_raw)
````

````python
# Título: Ejercicio 03 - Normalización de Ubicaciones y Patentes

def normalizar_cadenas(df: pd.DataFrame) -> pd.DataFrame:
  """
  Limpia las cadenas de texto en ubicaciones y patentes eliminando
  caracteres especiales y pasando a mayúsculas.

  Args:
    df (pd.DataFrame): DataFrame a limpiar.

  Returns:
    pd.DataFrame: DataFrame con strings normalizados.
  """
  df['ubicacion'] = df['ubicacion'].fillna('').str.replace(
    r'[^A-Za-z0-9 ]', '', regex=True
  ).str.upper()

  print("Primeras 10 ubicaciones:")
  print(df['ubicacion'].head(10))
  print("-" * 40)

  df['patente'] = df['patente'].fillna('').str.replace(
    r'[^A-Za-z0-9]', '', regex=True
  ).str.upper()

  df['patente'] = df['patente'].replace('', pd.NA)

  print("Últimas 10 patentes:")
  print(df['patente'].tail(10))

  return df


df_fines_raw = normalizar_cadenas(df_fines_raw)
````

````python
# Título: Ejercicio 03 - Tratamiento de Inconsistencias (Nulos y Outliers)

def tratar_inconsistencias(df: pd.DataFrame) -> pd.DataFrame:
  """
  Elimina filas con nulos en columnas relevantes y remueve outliers
  de velocidad utilizando el método IQR.

  Args:
    df (pd.DataFrame): DataFrame a limpiar.

  Returns:
    pd.DataFrame: DataFrame sin nulos críticos ni outliers.
  """
  cols_obs: List[str] = [
    'fecha',
    'patente',
    'velocidad_registrada',
    'velocidad_maxima'
  ]

  total_antes: int = len(df)
  df = df.dropna(subset=cols_obs)
  eliminados_nulos: int = total_antes - len(df)

  print(f"Se eliminaron {eliminados_nulos} filas.")
  print("Las columnas observadas son:")
  for col in cols_obs:
    print(f" - {col}")
  print("-" * 40)

  q1: float = df['velocidad_registrada'].quantile(0.25)
  q3: float = df['velocidad_registrada'].quantile(0.75)
  iqr: float = q3 - q1

  limite_inf: float = q1 - 1.5 * iqr
  limite_sup: float = q3 + 1.5 * iqr

  df_filtrado: pd.DataFrame = df[
    (df['velocidad_registrada'] >= limite_inf) &
    (df['velocidad_registrada'] <= limite_sup)
  ]

  eliminados_out: int = len(df) - len(df_filtrado)

  print(f"Se eliminaron {eliminados_out} filas.")
  print("Las columnas observadas son:")
  print(" - velocidad_registrada")

  return df_filtrado


df_fines_raw = tratar_inconsistencias(df_fines_raw)
````

````python
# Título: Ejercicio 03 - Cálculo de Excesos y Guardado Interim

def finalizar_procesamiento(df: pd.DataFrame) -> None:
  """
  Calcula excesos de velocidad, filtra no infractores y guarda
  el dataset procesado en la carpeta interim.

  Args:
    df (pd.DataFrame): DataFrame limpio y sin outliers.

  Returns:
    None: Guarda el dataset en disco.
  """
  df['exceso_velocidad_real'] = (
    df['velocidad_registrada'] - df['velocidad_maxima']
  )

  print("Primeras 10 patentes y su exceso real:")
  print(df[['patente', 'exceso_velocidad_real']].head(10))
  print("-" * 40)

  df['exceso_velocidad'] = (
    df['velocidad_registrada'] - (df['velocidad_maxima'] * 1.05)
  )

  print("Primeras 10 patentes y su exceso (margen 5%):")
  print(df[['patente', 'exceso_velocidad']].head(10))

  total_inicial: int = len(df)
  df_infractores: pd.DataFrame = df[df['exceso_velocidad'] > 0]
  eliminados: int = total_inicial - len(df_infractores)

  print(f"Se eliminaron {eliminados} filas.")

  ruta_interim: str = "urban_flow/data/interim/speeding_fines.csv"
  os.makedirs("urban_flow/data/interim", exist_ok=True)

  df_infractores.to_csv(ruta_interim, index=False)
  print(f"Dataset guardado en: {ruta_interim}")


finalizar_procesamiento(df_fines_raw)
````

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
# Título: Ejercicio 04 - Clase FineAnalyzer (POO)

class FineAnalyzer:
  """
  Clase encargada de realizar análisis estadísticos y rankings
  sobre el dataset de infracciones de velocidad.
  """

  def __init__(self, df: pd.DataFrame) -> None:
    """
    Inicializa la clase encapsulando una copia del DataFrame.

    Args:
      df (pd.DataFrame): Dataset limpio e intermedio de infracciones.

    Returns:
      None: Inicializa el objeto.
    """
    self.__df: pd.DataFrame = df.copy()

  def ranking_patentes(self, n: int = 5) -> pd.DataFrame:
    """
    Genera el ranking de las 'n' patentes con más multas.

        Args:
            n (int): Cantidad de patentes a mostrar en el ranking (default=5).

        Returns:
            pd.DataFrame: DataFrame con columnas 'patente' y 'cantidad',
            ordenado de mayor a menor e índice desde 1.
    """
    ranking: pd.DataFrame = (
      self.__df['patente']
      .value_counts()
      .head(n)
      .rename_axis('patente')
      .reset_index(name='cantidad')
    )

    ranking.index = range(1, len(ranking) + 1)
    return ranking

  def ranking_horarios(self) -> pd.DataFrame:
    """
    Genera el ranking de los 5 horarios con más multas.

    Returns:
      pd.DataFrame: DataFrame con columnas 'hora' y 'cantidad',
      ordenado de mayor a menor e índice desde 1.
    """
    ranking: pd.DataFrame = (
      self.__df['hora']
      .value_counts()
      .head(5)
      .rename_axis('hora')
      .reset_index(name='cantidad')
    )

    ranking.index = range(1, len(ranking) + 1)
    return ranking

  def promedio_exceso_velocidad(self) -> float:
    """
    Calcula el promedio del exceso de velocidad (con margen del 5%).

    Returns:
      float: Promedio del exceso de velocidad.
    """
    return float(self.__df['exceso_velocidad'].mean())

  def promedio_exceso_real(self) -> float:
    """
    Calcula el promedio del exceso de velocidad real (sin margen).

    Returns:
      float: Promedio del exceso real.
    """
    return float(self.__df['exceso_velocidad_real'].mean())

  def multas_por_ubicacion(self) -> pd.DataFrame:
    """
    Cuenta la cantidad de multas por ubicación.

    Returns:
      pd.DataFrame: DataFrame con columnas 'ubicacion' y 'cantidad',
      ordenado alfabéticamente por ubicación.
    """
    conteo: pd.DataFrame = (
      self.__df
      .groupby('ubicacion')
      .size()
      .reset_index(name='cantidad')
    )

    return conteo.sort_values(by='ubicacion')
````
